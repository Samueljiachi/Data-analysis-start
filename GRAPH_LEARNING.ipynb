```python
Codeplt.figure(figsize=(15,10))
#plt.figure(figure=(8,5))
sns.set_style("whitegrid") #changes the design of the graph
sns.set_context("paper") #makes your context smaller or bigger
sns.scatterplot(data = pen,x = "PROD_NAME",y = "TOT_SALES",hue = "DATE",)
sns.despine(left = True) #space between graph and background
```

```python
plt.figure(figure=(8,5))
sns.histplot(data = pen,x = "TOT_SALES",bins=10,kde=True,color = "red")
plt.title("Total sales of the year",fontsize = 14)
plt.xlabel("SALES")
plt.ylabel("x-axis")
plt.show()
```

```python
#REDUCING THE NUMBERS ON THE X AXIS SO IT STARTS FROM 0 AND END WITH 15
redo = pen[pen["TOT_SALES"]<15.0]
```

```python
sns.histplot(data = redo,x = "TOT_SALES")
```

```python
plt.figure(figure=(8,5))
sns.boxplot(data = redo,x="TOT_SALES")
```
